```python
import osmnx as ox

# Çalışma alanımızı Beyoğlu olarak güncelliyoruz
place_name = "Beyoğlu, Istanbul, Turkey"

print(f"{place_name} için harita verileri indiriliyor... Lütfen bekleyin.")

# OSM üzerinden sadece yaya yollarını bir ağ (network) topolojisi olarak çekiyoruz
G = ox.graph_from_place(place_name, network_type='walk')

print("Veri başarıyla çekildi!")
print(f"Sistemdeki Düğüm (Node - Kavşak/Nokta) sayısı: {len(G.nodes)}")
print(f"Sistemdeki Kenar (Edge - Sokak/Bağlantı) sayısı: {len(G.edges)}")
```

```python
import folium

print("Beyoğlu'ndaki potansiyel macera noktaları (POI) aranıyor...")

# Beyoğlu içindeki turistik noktaları, sanat eserlerini ve tarihi binaları çekiyoruz
tags = {'tourism': ['museum', 'artwork', 'gallery'], 'historic': ['monument', 'memorial']}
pois = ox.features_from_place(place_name, tags=tags)

# Haritamızın merkez koordinatlarını oluşturuyoruz
center_lat = pois.geometry.centroid.y.mean()
center_lon = pois.geometry.centroid.x.mean()
m = folium.Map(location=[center_lat, center_lon], zoom_start=14)

# Bulduğumuz noktaları haritaya kırmızı işaretçiler olarak ekliyoruz
for idx, row in pois.iterrows():
    if row.geometry.geom_type == 'Point':
        name = row.get('name', 'Gizemli Hedef')
        # İsimsiz noktaları (NaN) filtrelemek için küçük bir kontrol
        if str(name) == 'nan':
            name = 'Gizemli Hedef'

        folium.Marker(
            location=[row.geometry.y, row.geometry.x],
            popup=name,
            icon=folium.Icon(color="red", icon="info-sign")
        ).add_to(m)

print(f"Toplam {len(pois)} potansiyel hazine noktası bulundu!")

# Haritayı Colab ekranında göster
m
```

```python
import networkx as nx

# 1. Başlangıç noktamız (Örneğin Taksim Meydanı)
start_lat = 41.0369
start_lon = 28.9850

print("Ajan, hedefe giden en kısa uzamsal rotayı hesaplıyor...")

# 2. Hedef noktamız (Az önce bulduğumuz noktalardan ilkini seçiyoruz)
# Sadece nokta (Point) geometrisine sahip olanları filtreleyelim
point_pois = pois[pois.geometry.geom_type == 'Point']
target_poi = point_pois.iloc[0]
target_lat = target_poi.geometry.y
target_lon = target_poi.geometry.x
target_name = target_poi.get('name', 'Gizemli Hedef')

# 3. Bu koordinatlara en yakın yol ağındaki düğümleri (node) buluyoruz
orig_node = ox.distance.nearest_nodes(G, X=start_lon, Y=start_lat)
dest_node = ox.distance.nearest_nodes(G, X=target_lon, Y=target_lat)

# 4. En kısa yaya rotasını hesaplıyoruz
route = ox.shortest_path(G, orig_node, dest_node, weight='length')

# 5. Rotayı harita üzerinde çiziyoruz
m_route = folium.Map(location=[(start_lat + target_lat)/2, (start_lon + target_lon)/2], zoom_start=15)

# Başlangıç ve Bitiş işaretçileri
folium.Marker(
    [start_lat, start_lon],
    popup="Başlangıç Noktası",
    icon=folium.Icon(color="green", icon="user")
).add_to(m_route)

folium.Marker(
    [target_lat, target_lon],
    popup=target_name,
    icon=folium.Icon(color="red", icon="star")
).add_to(m_route)

# Rotayı (düğümlerin koordinatlarını alarak) haritaya mavi bir çizgi olarak ekliyoruz
route_coords = [(G.nodes[node]['y'], G.nodes[node]['x']) for node in route]
folium.PolyLine(locations=route_coords, color="blue", weight=5, tooltip="Ajanın Otonom Rotası").add_to(m_route)

print(f"Rota çizildi! Hedef: {target_name}")

# Haritayı göster
m_route
```

```python
import osmnx as ox
import networkx as nx
import folium
from google import genai
import random

# 1. API Ayarları (BURAYA KENDİ ANAHTARINI YAZMAYI UNUTMA!)
GOOGLE_API_KEY = "AIzaSyAEK29ImYN0srW7OJErlsdFcu7LWHcmZrk"
client = genai.Client(api_key=GOOGLE_API_KEY)

# 2. Çalışma Alanı ve Veri Çekimi
place_name = "Beyoğlu, Istanbul, Turkey"
print("1/4: Beyoğlu sokak ağı ve harita verileri çekiliyor... (Biraz sürebilir)")
G = ox.graph_from_place(place_name, network_type='walk')

tags = {'tourism': ['museum', 'artwork', 'gallery'], 'historic': ['monument', 'memorial']}
pois = ox.features_from_place(place_name, tags=tags)
point_pois = pois[pois.geometry.geom_type == 'Point']

# 3. Başlangıç ve Otonom Hedef Belirleme
start_lat = 41.0369 # Taksim Meydanı
start_lon = 28.9850

# Ajanımız rastgele bir gizli hedef seçiyor
hedef_index = random.randint(0, len(point_pois) - 1)
target_poi = point_pois.iloc[hedef_index]
target_lat = target_poi.geometry.y
target_lon = target_poi.geometry.x

# İsimsiz hedeflere standart bir isim atıyoruz
target_name = target_poi.get('name')
if str(target_name) == 'nan' or not target_name:
    target_name = 'Tarihi Gizemli Nokta'

print("2/4: Ajan hedefi belirledi ve uzamsal rotayı hesaplıyor...")

# 4. Uzamsal Rota Hesaplama (Routing Algorithm)
orig_node = ox.distance.nearest_nodes(G, X=start_lon, Y=start_lat)
dest_node = ox.distance.nearest_nodes(G, X=target_lon, Y=target_lat)
route = ox.shortest_path(G, orig_node, dest_node, weight='length')

# 5. Haritayı Çizme
m_final = folium.Map(location=[(start_lat + target_lat)/2, (start_lon + target_lon)/2], zoom_start=15)

# Yeşil işaretçi: Oyuncu, Siyah işaretçi: Hedef
folium.Marker([start_lat, start_lon], popup="Senin Konumun", icon=folium.Icon(color="green", icon="user")).add_to(m_final)
folium.Marker([target_lat, target_lon], popup="Gizli Hedef", icon=folium.Icon(color="black", icon="star")).add_to(m_final)

# Ajanın çizdiği mor rota
route_coords = [(G.nodes[node]['y'], G.nodes[node]['x']) for node in route]
folium.PolyLine(locations=route_coords, color="purple", weight=5, tooltip="Ajanın Otonom Yaya Rotası").add_to(m_final)

# 6. Bilmece Üretimi (LLM Orchestrator)
print("3/4: Ajan gizli hedef için bilmecesini hazırlıyor...")

prompt = f"""
Sen zeki, mistik bir aura taşıyan ve bir kahin gibi gizemli konuşan otonom bir 'Geocaching' oyun kurucu ajansın.
Oyuncunun kaderinde gitmesi gereken sıradaki gizli hedef: {target_name}.

Kurallar:
1. Oyuncuya asla doğrudan hedefin adını söyleme!
2. Hedefin tarihi, coğrafi veya fiziksel özelliklerini anlatan, içinde hafif şamanik semboller, yıldızlar veya kader temaları barındıran gizemli bir bilmece yaz.
3. Bilmece 2 veya 3 cümleyi geçmesin ve oyuncuyu o noktaya doğru yürümeye teşvik etsin.
"""

response = client.models.generate_content(
    model='gemini-2.5-flash',
    contents=prompt
)

print("4/4: İşlem tamamlandı!\n")
print("✨ AJANIN BİLMECESİ:")
print("--------------------------------------------------")
print(response.text)
print("--------------------------------------------------\n")

# Haritayı Colab ekranında göster
m_final
```
